# Task 04: Sets for Unique Vocabulary

**(a) Convert a list with duplicates into a set**

```python
raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]
vocab_set = set(raw)

print("raw:", raw)
print("vocab (set):", vocab_set)
print("Number of unique words:", len(vocab_set))

# Sets automatically remove duplicates because they are implemented as hash tables that
# only store one entry per unique hash value. When an item that's already present is
# added again, Python recognises the matching hash/equality and simply does not create
# a second entry, so every element in a set is guaranteed to be unique.
```

**(b) Union, intersection, difference of two sets**

```python
set_A = {"machine", "learning", "data", "model"}
set_B = {"deep", "learning", "neural", "data"}

union_set = set_A | set_B
intersection_set = set_A & set_B
difference_set = set_A - set_B

print("Union:", union_set)
print("Intersection:", intersection_set)
print("Difference (A - B):", difference_set)
```

**(c) Remove stopwords using set difference**

```python
stopwords = {"is", "the", "a", "in", "of", "and", "to"}
tokens_04c = ["nlp", "is", "a", "field", "of", "ai", "and", "the", "future"]

cleaned_tokens = set(tokens_04c) - stopwords
print("Cleaned tokens:", cleaned_tokens)
```

**(d) add(), discard(), and membership check**

# Task 06: String Cleaning Methods

**(a) strip(), lower(), replace() applied step by step**

```python
text = "   MACHINE Learning is AMAZING!!!   "

step1 = text.strip()
print("After strip():", repr(step1))

step2 = step1.lower()
print("After lower():", repr(step2))

step3 = step2.replace("!", "")
print("After replace('!', ''):", repr(step3))
```

**(b) split(), uppercase list comprehension, join()**

```python
words_06 = step3.split()
print("Tokenised words:", words_06)

uppercase_words = [w.upper() for w in words_06]
print("Uppercase words:", uppercase_words)

joined = " | ".join(uppercase_words)
print("Joined string:", joined)
```

**(c) replace() and find()**

```python
sentence_06c = "I love Java and Java is popular"

updated_sentence = sentence_06c.replace("Java", "Python")
first_python_pos = updated_sentence.find("Python")

print("Updated sentence:", updated_sentence)
print("First position of 'Python':", first_python_pos)
```

**(d) `normalize()` function**

```python
def normalize(text):
    return text.strip().lower().replace(",", "").replace(".", "")

sample1 = "  Hello, World.  "
sample2 = "NLP, is Amazing.   "

print("Cleaned sample1:", repr(normalize(sample1)))
print("Cleaned sample2:", repr(normalize(sample2)))
```

# Task 07: String Validation Methods

**(a) startswith(), endswith(), find()**

```python
text_07 = "NLP is the future of AI research"

print("Starts with 'NLP'?", text_07.startswith("NLP"))
print("Ends with 'research'?", text_07.endswith("research"))
print("Position of 'future':", text_07.find("future"))
```

**(b) Classify each token as alphabetic, numeric, or mixed**

```python
tokens_07b = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens_07b:
    if token.isalpha():
        print(f"{token}: purely alphabetic")
    elif token.isdigit():
        print(f"{token}: purely numeric")
    else:
        print(f"{token}: mixed")
```

**(c) Separate words into three groups**

```python
words_07c = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for w in words_07c:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)

print("pure_words:", pure_words)
print("pure_numbers:", pure_numbers)
print("mixed_tokens:", mixed_tokens)
```

**(d) `validate_token()` function**

```python
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

for token in ["chat", "99", "gpt4", "bert", "2024", "nlp!"]:
    print(f"{token}: {validate_token(token)}")
```

# Task 08: Functions and Lambda for NLP Pipelines

**(a) `clean()` function vs. equivalent lambda**

```python
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

sample_08a = "  HELLO Natural Language Processing  "

print("clean():", repr(clean(sample_08a)))
print("clean_lambda():", repr(clean_lambda(sample_08a)))
```

**(b) `tokenize()` and `remove_short()` chained together**

```python
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

sentence_08b = "I love working with NLP tools in Python daily"
result_08b = remove_short(tokenize(sentence_08b), 3)

print("Tokens longer than 3 characters:", result_08b)
```

**(c) `word_lengths()` function**

```python
def word_lengths(tokens):
    return {t: len(t) for t in tokens}

tokens_08c = ["language", "model", "ai", "processing", "deep"]
print("Word lengths:", word_lengths(tokens_08c))
```

**(d) Full `nlp_pipeline()` combining clean, tokenize, remove_short, word_lengths**

```python
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

sentence_a = "  Natural Language Processing enables machines to understand text  "
sentence_b = "Deep learning models require large amounts of data"

print("Pipeline on sentence_a:", nlp_pipeline(sentence_a, 3))
print("Pipeline on sentence_b:", nlp_pipeline(sentence_b, 3))
```

# Task 09: map(), filter(), sorted() and Counter

**(a) map() with lambda: lowercase and character length**

```python
words_09 = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lowercased = list(map(lambda w: w.lower(), words_09))
lengths = list(map(lambda w: len(w), words_09))

print("Lowercased:", lowercased)
print("Lengths:", lengths)
```

**(b) filter() with lambda: long words and words containing "t"**

```python
words_long = list(filter(lambda w: len(w) > 3, words_09))
words_with_t = list(filter(lambda w: "t" in w.lower(), words_09))

print("Words with more than 3 characters:", words_long)
print("Words containing 't':", words_with_t)
```

**(c) sorted() three different ways**

```python
alphabetical = sorted(words_09)
by_length_asc = sorted(words_09, key=len)
by_length_desc = sorted(words_09, key=len, reverse=True)

print("Alphabetical:", alphabetical)
print("By length (ascending):", by_length_asc)
print("By length (descending):", by_length_desc)
```

**(d) Counter: most common words, frequency lookup, vocabulary size**

```python
sentence_09d = "nlp nlp python ai nlp python text ai nlp python text"
words_list = sentence_09d.split()
freq = Counter(words_list)

print("Counter:", freq)
print("3 most common words:", freq.most_common(3))
print("Frequency of 'python':", freq["python"])
print("Total number of words:", len(words_list))
print("Size of unique vocabulary:", len(freq))
```
